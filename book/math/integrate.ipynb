标准正态分布的概率密度函数 (PDF) 和累积分布函数 (CDF)

```python
import scipy.stats as stats

# 定义标准正态分布 (均值 μ=0, 标准差 σ=1)
# 如果你的数据不是标准正态分布，可以用 loc=μ, scale=σ 参数调整
norm_dist = stats.norm(loc=0, scale=1)

# 定义要计算的倍数
sigma_levels = [1, 2, 3]

print("正态分布区间概率 (标准正态分布 μ=0, σ=1):")
print("-" * 30)

for k in sigma_levels:
    # 计算区间 (μ - kσ, μ + kσ) 的累积概率
    # cdf 是累积分布函数，计算从 -∞ 到 x 的概率
    prob = norm_dist.cdf(k) - norm_dist.cdf(-k)
    print(f"P(μ ± {k}σ) = {prob:.6f}  (理论值: { [0.6827, 0.9545, 0.9973][k-1] })")
```

菲涅尔镜（Fresnel Lens，螺纹透镜）本质是：把传统厚凸透镜 “拍扁、挖空”，只保留每层的折射斜面，做到极薄、超轻、大口径、低成本，聚光效果接近厚透镜。


```python

import scipy.integrate as integrate
import scipy.special as special
result = integrate.quad(lambda x: special.jv(2.5,x), 0, 4.5)


from numpy import sqrt, sin, cos, pi
I = sqrt(2/pi)*(18.0/27*sqrt(2)*cos(4.5) - 4.0/27*sqrt(2)*sin(4.5) +
                sqrt(2*pi) * special.fresnel(3/sqrt(pi))[0])

print(abs(result[0]-I))
```
