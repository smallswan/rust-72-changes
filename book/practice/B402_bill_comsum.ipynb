```python
import pyecharts.options as opts
from pyecharts.charts import Line
import pandas as pd
import numpy as np

month_end = pd.date_range("2016-01-31", "2025-05-31", freq="ME")
xaxis_data = month_end.strftime("%Y-%m-%d").tolist()
payable = pd.read_excel("C:\\repositories\\rust\\rust-72-changes\\src\\practice\\payable.xlsx",header=0, index_col=0, na_values=['',' ', 'NA', 'NULL'])


payable_values = payable.round(2).apply(pd.to_numeric, downcast='float', errors='coerce').fillna(0)

print(payable_values)
# print(np.nan_to_num(payable_values, nan=0.0))
print(f"数据形状: {payable_values.shape}")
print(f"行数: {payable_values.shape[0]}, 列数: {payable_values.shape[1]}")
# 中位数（按列）
median_values = payable_values.median(axis=0)
print(f"中位数: {median_values}")
print(f"最后一列的中位数:\n{payable_values.iloc[64:67,-1]}")

nums = pd.DataFrame([55, 62, 68, 72, 75, 78, 81, 85, 88, 92, 95, 98])
print(nums.describe())

received = pd.read_excel("C:\\repositories\\rust\\rust-72-changes\\src\\practice\\received.xlsx")
received_B402 = received.iloc[-6, :].apply(pd.to_numeric, downcast='float', errors='coerce').fillna(0).values.cumsum()

line_chart = Line()

# 获取实际的月份标签（根据数据列数）
num_months = payable_values.shape[1]
actual_xaxis = xaxis_data[-num_months:]  # 取与数据匹配的最近几个月

line_chart.add_xaxis(xaxis_data=actual_xaxis)

line_chart.set_global_opts(
    tooltip_opts=opts.TooltipOpts(
        # 使用 JsCode 进行更复杂的自定义
        formatter=opts.TooltipOpts(
            formatter="{c:.2f}"  # 简化写法，将系列数据格式化为两位小数
        )
    )
)

# 遍历每一行（每个住户）
# payable_values.shape[0]
for row_idx in range(60):
    #resident_name = f"住户{row_idx + 1}"
    resident_name = payable_values.index[row_idx]
    cumulative_payable = payable_values.iloc[row_idx, :].cumsum()
    y_axis_data = cumulative_payable.values.round(2).tolist()
    line_chart.add_yaxis(
        series_name=resident_name,
        y_axis=y_axis_data,
        label_opts=opts.LabelOpts(is_show=False),
        markpoint_opts=opts.MarkPointOpts(
            data=[
                opts.MarkPointItem(type_="max", name="最大值"),
                opts.MarkPointItem(type_="min", name="最小值"),
            ]
        ),
        markline_opts=opts.MarkLineOpts(
            data=[opts.MarkLineItem(type_="average", name="平均值")]
        ),
    )

(
    line_chart
    .set_global_opts(
        title_opts=opts.TitleOpts(title="所有住户累计应收款项", subtitle="纯属虚构"),
        tooltip_opts=opts.TooltipOpts(trigger="axis"),
        toolbox_opts=opts.ToolboxOpts(is_show=False),
        xaxis_opts=opts.AxisOpts(type_="category", boundary_gap=False),
        legend_opts=opts.LegendOpts(type_="scroll", pos_top="10%", pos_left="center"),
    )
    .render("all_residents_payable_chart.html")
)
```
