
## 中文分词
jieba 是 Python 生态中最流行的中文分词库，号称"最好的 Python 中文分词组件"，在 GitHub 上拥有超过 2.4 万个 Star。
### jieba三种分词方式
1.精确模式
jieba.cut(text, cut_all=False)
2.全模式
jieba.cut(text, cut_all=True)
3.搜索引擎模式
jieba.cut_for_search(text)


```python
import jieba

# 1. 基本分词（返回迭代器，推荐用于大文本）
words = jieba.cut("我爱自然语言处理", cut_all=False)  # 精确模式
print("/".join(words))  # 我/爱/自然语言/处理

# 2. 返回列表（日常使用最方便）
words = jieba.lcut("我爱自然语言处理")  # lcut = list + cut
print(words)  # ['我', '爱', '自然语言', '处理']

# 3. 搜索引擎模式
words = jieba.cut_for_search("清华大学位于北京")
print("/".join(words))  # 清华/大学/清华大学/位于/北京

# 4. 添加自定义词典（解决新词、专业术语识别问题）
#jieba.load_userdict("my_dict.txt")  # 每行格式：词语 词频 词性
```

```python
import jieba
from sklearn.feature_extraction.text import TfidfVectorizer

def chinese_tokenizer(text):
    """自定义分词函数"""
    return list(jieba.cut(text))

corpus = [
    '这双鞋子真好看',
    '这个衣服质量不错',
    '鞋子真好看，我喜欢这双鞋子'
]

# 使用自定义分词器
#vectorizer = TfidfVectorizer(tokenizer=chinese_tokenizer, token_pattern=None)
# 方式二：更简洁的写法，直接指向 jieba.lcut 方法
vectorizer = TfidfVectorizer(tokenizer=jieba.lcut, token_pattern=None)
tfidf_matrix = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(vectorizer.idf_)
print(vectorizer.vocabulary_)

print(tfidf_matrix.toarray())
```
